# Cell 对象
`atoms.cell`是一个Cell对象。

see [Class ase.cell.Cell(array)](../ase/cell.py)

## 1.Cell对象的属性

- rank：int -->维度

- orthorhombic:bool-->返回cell是否被对角矩阵表示

- headedness: int -->手性，表示cell行列式符号，1对应右手，-1对应左手。


## 2.Cell对象的方法

### 2.1 实例方法

#### 2.1.1 参数类：

- angles():返回一个数组，带有三个角度。

- lengths():返回每个晶格矢量的长度的数组。

- mask():晶格长度非0的布尔数组。

- area(i)： i轴对应的2个矢量张成的面积。

- areas():（1,2）、（2,0）,(0,2)组成的面积。

- normal(i)：某个索引i的法向量(由i之外的两个轴构成)。

- normals():类似areas()

- cellpar(radian=False):返回长度为[6,1]格式的cell参数列表。

##### 倒空间相关：
- get_bravis_lattice(eps=0.0002,pbc=True):获取该cell的[BravaisLattice](../ase/lattice/__init__.py#27)实例，

- reciprocal()->cell：
>获得cell对象在机器精度下满足:`cell.reciprocal() @ cell.T == np.diag(cell.mask())`，注意不包括2pi因子。

- scaled_positions



#### 2.1.2 功能类：

##### 约化：
- minkowski_resduce()：对该cell进行Minkowski约化，返回新的cell。
- niggli_reduce(eps=1e-05)

---

- permute_axes(permutation)：
例如：
```python
>>>cell=Cell.fromcellpar([2,3,4])
>>>cell
>>>Cell([2.0, 3.0, 4.0])
>>>cell.permute_axes([0,2,1])
>>>cell
>>> cell([2，4,3])
```

##### 高级：

- **bandpath**(path:str,npoints:int,density:float,special points:dict,eps:float,pbc:bool):
> see [bandpath()](../ase/cell/cell.py#131),为Cell创建一个新类[ase.dft.kpoints.BandPath](../ase/dft/kpoints.py),用于存放电子能带计算的k-points的列表。


### 2.2 类方法

- ascell(cell)：返回一个Cell对象。

- new(cell=None)：
 -- cell是3个数字时，假定是三个长度和直角。
 -- cell是6个数字，假定是3长3角。
 -- cell=3*3 matrix时对应向量。
