# GPAW


- GPAW是一个被部署成python库的C后端的电子结构代码。

- GPAW目前只发布了Linux和macOS的安装包，在Windows上暂时没有支持。

- 它包含一些命令行工具，但是通过配合ASE一同使用。

- 由于这种紧密的关系，GPAW的开发者往往也是ASE的开发者。

- GPAW在实现投影增强波方法（PAW）的时候需要依赖原子的贋势配置。


```python
from gpaw import GPAW
from ase.build import bulk
atoms= bulk('Cu', 'fcc', a=3.6)
```

下面我们配置GPAW作为一个calculator，需要设定一些参数:

- xc：交换关联泛函。

-  kpts: Briilouin -zone的采样。

- mode: 设置平面波基矢集合。


```python
atoms.calc=GPAW(xc='PBE',kpts=(3,3,3),mode=PW(400))
```
